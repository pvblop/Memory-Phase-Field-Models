```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
f = lambda psi, R: 1/4 * (psi ** 2 - 1) ** 2 - R * psi * (1 - psi / 2)

psi = np.linspace(-1.5, 1.5, 100)

fig, axs = plt.subplots(1, 3, figsize=(15, 5))


ax = axs[0]
R = 0.0
F = f(psi, R)
ax.plot(psi, F, label=r"$F(\psi)$", color="black")
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_position(('data', 0))

ax.set_xticks([-1, 0, 1])
ax.set_yticks([])

ax.tick_params(axis='x', width=2, length=8, labelsize=15)
\

for spine in ax.spines.values():
    spine.set_linewidth(2)


ax = axs[1]
R = 0.15
F = f(psi, R)
ax.plot(psi, F, label=r"$F(\psi)$", color="black")
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_position(('data', 0))

ax.set_xticks([-1, 0, 1])
ax.set_yticks([])
ax.tick_params(axis='x', width=2, length=8, labelsize=15)

for spine in ax.spines.values():
    spine.set_linewidth(2)


ax = axs[2]
R = 0.45
F = f(psi, R)
ax.plot(psi, F, label=r"$F(\psi)$", color="black")
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_position(('data', 0))

ax.set_xticks([-1, 0, 1])
ax.set_yticks([])
ax.tick_params(axis='x', width=2, length=8, labelsize=15)


for spine in ax.spines.values():
    spine.set_linewidth(2)

plt.savefig("free_energy_gauss.svg", dpi=600, bbox_inches="tight")
```
