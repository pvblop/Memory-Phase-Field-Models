```python
import numpy as np
import matplotlib.pyplot as plt
from operators import beads_dist
from scipy.optimize import brentq
```

```python
Lx, Ly = 50, 50
Nx, Ny = 100, 100

x = np.linspace(0, Lx, Nx, endpoint=False)
y = np.linspace(0, Ly, Ny, endpoint=False)
dx = Lx / Nx
dy = Ly / Ny
X, Y = np.meshgrid(x, y, indexing="xy")


def num_bd(rbm, sigma, x0):
    Nb = beads_dist("pol", rbm, X, Y, x0, 25, sigma)
    num_bd = np.sum(Nb) * dx * dy
    return num_bd


num = num_bd(0.26, 25, 25)
print(num)


aux = lambda rbm: num_bd(rbm, 10, 12.5) - num
rbm_solution = brentq(aux, 0.1, 1.0)
print(rbm_solution)

print(num_bd(rbm_solution, 10, 12.5))
```
