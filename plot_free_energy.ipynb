```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# -----------------------------
# Malla paramétrica tipo gnuplot
# -----------------------------
nu, nv = 100, 100
u = np.linspace(-np.pi, np.pi, nu)
v = np.linspace(0.0, 1.3, nv)

U, V = np.meshgrid(u, v)

# Coordenadas del plano
X = np.cos(U) * V
Y = np.sin(U) * V

Q = V * (np.cos(0) + np.sin(0)) 

# -----------------------------
# TU FUNCIÓN VA AQUÍ
# -----------------------------
# Ejemplo cualquiera:
Z = 0.25*(V**2 - 1)**2 - 0.5 * V * (V - Q)

# -----------------------------
# Figura
# -----------------------------
fig = plt.figure(figsize=(3, 3), dpi=200)
ax = fig.add_subplot(111, projection='3d')

# Superficie principal
ax.plot_wireframe(
    X, Y, Z,
    rcount=nu//12, ccount=nv//12,
    linewidth=0.7,
    color='black',
)


# -----------------------------
# Estilo tipo conferencia
# -----------------------------
ax.view_init(elev=35, azim=253)   # parecido a set view 52,253
ax.set_box_aspect((1, 1, 0.7))

# Sin ticks
ax.set_xticks([])
ax.set_yticks([])
ax.set_zticks([])

# Sin etiquetas
ax.set_xlabel(r'')
ax.set_ylabel(r'')
ax.set_zlabel(r'')

# Quitar paneles
ax.xaxis.pane.fill = False
ax.yaxis.pane.fill = False
ax.zaxis.pane.fill = False

# Quitar bordes de los paneles
ax.xaxis.pane.set_edgecolor((1, 1, 1, 0))
ax.yaxis.pane.set_edgecolor((1, 1, 1, 0))
ax.zaxis.pane.set_edgecolor((1, 1, 1, 0))

# Quitar líneas del cubo 3D
ax.xaxis.line.set_color((1, 1, 1, 0))
ax.yaxis.line.set_color((1, 1, 1, 0))
ax.zaxis.line.set_color((1, 1, 1, 0))

# Sin grilla
ax.grid(False)

# Ajustar límites verticales para dejar espacio a los contornos base

plt.tight_layout()
plt.show()
```

```python
px = np.linspace(-1.5, 1.5, 200)
py = np.linspace(-1.5, 1.5, 200)
PX, PY = np.meshgrid(px, py)
P = np.sqrt(PX ** 2 + PY ** 2)


f = 0.25*(P**2 - 1)**2 + 0.1 * ((PX - 1) * PX + (PY) ** 2)

plt.imshow(f, extent=(-1.5, 1.5, -1.5, 1.5), origin='lower', cmap='magma', clim=(-0.0, 0.30))
plt.colorbar(label=r'$f(\mathbf{p})$')
plt.xlabel(r'$p_x$')
plt.ylabel(r'$p_y$')
plt.tight_layout()
plt.savefig('free_energy_contour_q.png', dpi=600)
```
